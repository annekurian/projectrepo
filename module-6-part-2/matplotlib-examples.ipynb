```python
import pandas as pd

data = {
    'Name':['Bob', 'Jessica', 'Mary', 'John'],
    'Grade':[75, 69, 77, 98],
    'Gender':['Male','Female','Female','Male'],
    'Status':['Senior', 'Junior', 'Senior', 'Senior']
}

df = pd.DataFrame(data)
df.describe()
df.boxplot(column='Grade')
```

```python
data = {
    'Name':['Bob', 'Jessica', 'Mary', 'John', 'Charles'],
    'Grade':[75, 69, 77, 98, 130],
    'Gender':['Male','Female','Female','Male', 'Male'],
    'Status':['Senior', 'Junior', 'Senior', 'Senior', 'Junior']
}

df = pd.DataFrame(data)
df.describe()
df.boxplot(column='Grade')
```
