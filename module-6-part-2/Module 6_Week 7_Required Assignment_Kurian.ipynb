```python
import numpy as np

np.random.seed(42)
nparray = np.random.randn(3,3)
print(nparray)
```

```python
nparrayA = np.round(nparray, decimals=2)
print(nparrayA)
print('Data type of the array: ', nparrayA.dtype)
```

```python
nparray_add = nparrayA+10
print(nparray_add)
```

```python
print(f'Data type of array added with value 10: {nparray_add.dtype}')
print(f'Shape of array added with value 10: {nparray_add.shape}')
print(f'Size of array added with value 10: {nparray_add.size}')
```

```python
nparrayB = np.ones((3,3))
print(nparrayB)
```

```python
nparray_mul = nparrayA * nparrayB
print(nparray_mul)
```

```python
print(f'Data type of element-wise multiplication result: {nparray_mul.dtype}')
print(f'Shape of element-wise multiplication result: {nparray_mul.shape}')
print(f'Size of element-wise multiplication result: {nparray_mul.size}')
```

```python
print('Task - 2: Pandas')
print('-'*20)
import pandas as pd
pdata1 = {
    'Name':['Arthur', 'Jane', 'Jacob', 'Alice', 'Bob'],
    'EmployeeId':[101, 102, np.nan, np.nan, 112],
    'Role':['Manager', None,'Analyst' ,None, 'Developer']
}
frame = pd.DataFrame(pdata1, index=['us', 'uk', 'ca', 'in', 'au'])
print(frame)
```

```python
frame.isnull()
```

```python
dropped_frame = frame.dropna()
print(dropped_frame)
```

```python
values={'EmployeeId':100, 'Role':'TBA'}
filled_frame = frame.fillna(value=values)
print(filled_frame)
```

```python
print('Task - 3: Matplotlib')
print('-'*20)
import matplotlib.pyplot as plt

hour_series = np.array([f"{h:02d}:00" for h in range(6, 20,2)])
print(hour_series)
```

```python
np.random.seed(42)
temperature_series = np.random.uniform(26,32, 7)
print(np.round(temperature_series, 2))
```

```python
plt.plot(hour_series, temperature_series)
plt.title('Today\'s Temperature')
plt.xlabel('Hour')
plt.ylabel('Temperature')
```

```python
plt.plot(hour_series, temperature_series, color='green', marker='*', markeredgecolor='r', linewidth=1)
```
